### Perform data quality checks

```python
# Missing values
missing = pd.DataFrame({
    "Missing_Count": dataset.isnull().sum(),
    "Missing_Percentage": dataset.isnull().mean().mul(100).round(2)
})


display(missing.sort_values("Missing_Percentage", ascending=False))
```

```python
# Duplicate records
# Entire duplicate rows
print("Duplicate rows:", dataset.duplicated().sum())

# Duplicate transaction IDs
print(
    "Duplicate transaction IDs:",
    dataset["transaction_id"].duplicated().sum()
)
```

```python
# Check dates and numerical values
# Convert the date column to a datetime type
dataset["transaction_date"] = pd.to_datetime(
    dataset["transaction_date"], errors="coerce"     # there are three options for error=coerce , error=rise, error=ignore
)

print("Invalid dates:", dataset["transaction_date"].isna().sum())

# Check important numerical columns
display(
    dataset[["transaction_amount", "account_balance",
        "credit_score", "emi_amount"]].describe().T
)
```

### Clean and prepare the data

```python
# Standardize column names
dataset.columns = dataset.columns.str.strip().str.lower()

print(dataset.columns.tolist())
```

```python
# Check categorical values
categorical_cols = [
    "account_type",
    "transaction_type",
    "transaction_direction",
    "merchant_category",
    "transaction_status",
    "channel",
    "kyc_status"
]

for col in categorical_cols:
    print(f"\n{col}")
    print(dataset[col].value_counts(dropna=False))
```

```python
# Create useful analytical columns
# Extract date-related information
dataset["year"] = dataset["transaction_date"].dt.year
dataset["month"] = dataset["transaction_date"].dt.month
dataset["year_month"] = dataset["transaction_date"].dt.to_period("M").astype(str)
dataset["day_of_week"] = dataset["transaction_date"].dt.day_name()

# Create transaction amount bands
dataset["amount_band"] = pd.cut(
    dataset["transaction_amount"],
    bins=[-float("inf"), 1000, 10000, 100000, float("inf")],
    labels=["Up to 1K", "1K-10K", "10K-1L", "Above 1L"]
)

# Display the new columns
display(dataset[[
    "transaction_date",
    "transaction_amount",
    "year",
    "year_month",
    "day_of_week",
    "amount_band"
]].head())
```

### Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

### Questions

```python
# 1) What is the overall transaction activity?
total_transactions = len(dataset)
unique_customers = dataset["customer_id"].nunique()
total_transaction_value = dataset["transaction_amount"].sum()
average_transaction_value = dataset["transaction_amount"].mean()
median_transaction_value = dataset["transaction_amount"].median()

print("Total transactions:", f"{total_transactions:,}")
print("Unique customers:", f"{unique_customers:,}")
print("Total transaction value:", f"₹{total_transaction_value:,.2f}")
print("Average transaction value:", f"₹{average_transaction_value:,.2f}")
print("Median transaction value:", f"₹{median_transaction_value:,.2f}")
```

```python
# 2) Which payment methods are most popular?
payment_summary = (
    dataset.groupby("transaction_type")
      .agg(
          transaction_count=("transaction_id", "count"),
          total_value=("transaction_amount", "sum"),
          average_value=("transaction_amount", "mean")
      )
      .sort_values("transaction_count", ascending=False)
)

display(payment_summary.round(2))
```

```python
payment_summary["transaction_count"].plot(
    kind="bar",
    figsize=(10, 5),
    title="Transaction Count by Payment Method"
)

plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 3) How does transaction activity change over time?
monthly_summary = (
    dataset.groupby("year_month")
      .agg(
          transaction_count=("transaction_id", "count"),
          total_value=("transaction_amount", "sum")
      )
      .sort_index()
)

monthly_summary["transaction_count"].plot(
    figsize=(14, 5),
    title="Monthly Transaction Volume"
)

plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# 4) Which merchant categories account for the highest spending?
merchant_summary = (
    dataset.groupby("merchant_category")
      .agg(
          transaction_count=("transaction_id", "count"),
          total_value=("transaction_amount", "sum"),
          average_value=("transaction_amount", "mean")
      )
      .sort_values("total_value", ascending=False)
)

display(merchant_summary.round(2))
```

```python
top_merchants = merchant_summary.head(10).sort_values("total_value")

top_merchants["total_value"].plot(
    kind="barh",
    figsize=(10, 6),
    title="Top 10 Merchant Categories by Transaction Value"
)

plt.xlabel("Total Transaction Value (₹)")
plt.ylabel("Merchant Category")
plt.tight_layout()
plt.show()
```

### Analyze fraud indicators

```python
# 5) What percentage of transactions are labelled as fraudulent?
fraud_summary = (
    dataset.groupby("is_fraud")
      .agg(
          transaction_count=("transaction_id", "count"),
          average_amount=("transaction_amount", "mean"),
          total_amount=("transaction_amount", "sum")
      )
)

fraud_summary["percentage_of_transactions"] = (
    fraud_summary["transaction_count"] / len(dataset) * 100
)

display(fraud_summary.round(2))
```

```python
# Compare transaction amounts for the two groups:
sns.boxplot(
    data=dataset,
    x="is_fraud",
    y="transaction_amount"
)

plt.yscale("log")
plt.title("Transaction Amount Distribution by Fraud Label")
plt.xlabel("Fraud Label (0 = No, 1 = Yes)")
plt.ylabel("Transaction Amount (log scale)")
plt.tight_layout()
plt.show()
```

```python
# compare fraud labels by payment method, transaction hour, channel, and transaction status.
fraud_by_method = pd.crosstab(
    dataset["transaction_type"],
    dataset["is_fraud"],
    normalize="index"
) * 100

display(fraud_by_method.round(2))
```
